```octave
n = 5
```

```octave
A = zeros(2*n);
tic
for ii = n:-1:1
    A(ii, :) = ones(1, 2*n)*ii;
    A(2*n - ii + 1, :) = ones(1, 2*n)*ii;
    A(:, ii) = ones(2*n, 1)*ii;
    A(:, 2*n - ii + 1) = ones(2*n, 1)*ii;
end
toc

A
```

```octave
A = zeros(2*n);
o = ones(1, 2*n);

tic
for ii = n:-1:1
    A(ii, :) = o*ii;
    A(2*n - ii + 1, :) = o*ii;
    A(:, ii) = o.*ii;
    A(:, 2*n - ii + 1) = o.*ii;
end
toc

A
```
